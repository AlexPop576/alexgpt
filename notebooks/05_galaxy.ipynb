```python
import sys; sys.path.append("..")
from alexgpt.bpe import BPETokenizer
tokenizer = BPETokenizer.load("../checkpoints/galaxy_tokenizer.json")
```

```python
text = "I love my flurbix"

def show_tokens(tokenizer, text):
    ids = tokenizer.encode(text)
    print("|", end="")
    for id in ids:
        print(f"{tokenizer.pieces[id]} |", end="")

show_tokens(tokenizer, text)
```
